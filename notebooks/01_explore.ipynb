```python
import pandas as pd
from pathlib import Path

DATA = Path("../data")
players = pd.read_csv(DATA / "players.csv")
vals = pd.read_csv(DATA / "player_valuations.csv", parse_dates=["date"])

top5 = ["GB1", "ES1", "L1", "IT1", "FR1"]

print("Valuation date range:", vals.date.min(), "->", vals.date.max())
recent = vals[(vals.player_club_domestic_competition_id.isin(top5)) & (vals.date >= "2021-07-01")]
print("Valuations in top 5 leagues since 2021:", len(recent))
print("Unique players:", recent.player_id.nunique())
print()
print(players.position.value_counts())
print()
print(vals.market_value_in_eur.describe())
```
